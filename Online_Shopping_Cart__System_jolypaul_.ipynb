#Online Shopping Cart System

```python
#Part A – Class Creation
class ShoppingCart():
  def __init__(self, customer_name, items=None):
      self.customer_name=customer_name
      if items==None:
        self.items=[]
      else:
        self.items =items

#Part B – Core Methods Implementation

  def add_item(self ,price):
      if price> 0:
         self.items.append(price)

  def remove_item(self,price):
      if price in self.items:
         self.items.remove(price)

      else:
        print("Item not found in the cart")
  def calculate_total(self):
      total_price=0
      for i in self.items:
          total_price+=i
          #i+=1
      return total_price

  def apply_discount(self,total):
    if total<3000:
       #print("not discount")
       return total

    else:
      total = total-(total*0.1)
      return total

#Part C – Display & Utility Methods

  def display_cart(self):

    print(f"Customer name:-{self.customer_name}")
    print(f"Items in the cart:-{self.items}")



    total = self.calculate_total()
    print(f"Total amount:-{total}")

    print(f"Final amount after discount:-{self.apply_discount(total)}")
    print(f"Number of items in the cart:{len(self.items)}")

  def clear_cart(self):
    self.items.clear()
    print("All items removed.")

  #Part D – Object Usage & Output
  def __add__(self,other):
    new_customer = self.customer_name + " & " + other.customer_name
    new_items = self.items+other.items
    return ShoppingCart(new_customer,new_items)


cart = ShoppingCart("joly paul",[500,300,600])
cart.add_item(600)
    #print(cart.items)

cart.remove_item(600)
    #print(cart.items)
cart.display_cart()
cart.clear_cart()

#Part D – Object Usage & Output

cart1 = ShoppingCart("Joly paul", [500, 300])

cart1.add_item(1000)
cart1.remove_item(300)

cart2 = ShoppingCart("poly paul", [700, 1500])

cart2.add_item(1200)
cart2.remove_item(700)

print("\nCart 1 Details:")
cart1.display_cart()


print("\nCart 2 Details:")
cart2.display_cart()


combined_cart = cart1 + cart2

print("\nCombined Cart Details:")
combined_cart.display_cart()

print()
combined_cart.display_cart()


```
